```python
def main():

    Path("data/processed").mkdir(parents=True, exist_ok=True)
    RAW_PATH = "data/raw/rawdata.csv"
    PAPER_PATH = "data/processed/papers.csv"
    AUTHOR_PATH = "data/processed/paper_author.csv"
    df = pd.read_csv(
        RAW_PATH,
        keep_default_na=False
    )

    print("Raw rows:", len(df))
    print("Unique papers:", df["work_id"].nunique())


    paper_columns = [
        "work_id",
        "title",
        "abstract",
        "topics",
        "keywords",
        "publication_year"
    ]

    papers = df[paper_columns].drop_duplicates(
        subset="work_id"
    )

    papers.to_csv(
        PAPER_PATH,
        index=False
    )


    authors = df[
        ["work_id", "author_id", "author_name"]
    ].copy()

    # Remove records where author ID is missing
    authors = authors[
        authors["author_id"] != ""
    ]

    # Remove accidental duplicate relationships
    authors = authors.drop_duplicates(
        subset=["work_id", "author_id"]
    )

    authors.to_csv(
        AUTHOR_PATH,
        index=False
    )



    print("\n--- Dataset Statistics ---")

    print("Paper records:", len(papers))
    print("Paper-author records:", len(authors))

    print(
        "Unique researchers:",
        authors["author_id"].nunique()
    )

    print(
        "Missing abstracts:",
        (papers["abstract"] == "").sum()
    )

    print(
        "Missing author IDs in raw data:",
        (df["author_id"] == "").sum()
    )

    print(
        "Duplicate paper IDs in raw data:",
        df["work_id"].duplicated().sum()
    )

    print("\nSaved:")
    print(PAPER_PATH)
    print(AUTHOR_PATH)


if __name__ == "__main__":
    main()
```
